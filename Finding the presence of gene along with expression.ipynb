```python
#Project 1- Find the presence of MYC gene along with the maximum expression
genes=["BRCA1","TP53","EGFR","MYC"]
expression=[25,80,45,60]
if "MYC" in genes:
  print("MYC is present")
else:
  print("MYC is absent")
maximum=max(expression)
print("maximum expression:",maximum)
i=expression.index(max(expression))
print("Highest expression gene:",genes[i])
```
