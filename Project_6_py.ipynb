```python
%pip install biopandas
import pandas as pdb
from biopandas.pdb import PandasPdb
import os
from google.colab import drive
drive.mount('/content/drive')
path="/content/drive/My Drive/2v8t.pdb"
ps=PandasPdb().read_pdb(path)
print(ps)
df=ps.df['ATOM']
df.columns=['A1','A2','A3','A4','A5','A6','A7','A8','A9','A10','A11','A12','A13','A14','A15','A16','A17','A18','A19','A20','A21']
#Concatinating the residue name and residue number
df["A22"]=df['A6'].astype(str)+'_'+df['A9'].astype(str)
print(df)
#Creating subset for both A and B chain
subsetA=df[df['A8']=='A']
print(subsetA)
subsetB=df[df['A8']=='B']
print(subsetB)
#Finding the number of unique amino acids from A and B chain
countA=subsetA['A6'].value_counts()
print(countA)
countB=subsetB['A6'].value_counts()
print(countB)
#Finding the unique amino acids fround in conactinated column
uniqueA=subsetA['A22'].unique()
print(uniqueA)
uniqueB=subsetB['A22'].unique()
print(uniqueB)
```
