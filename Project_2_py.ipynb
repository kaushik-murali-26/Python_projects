```python
#Project 2 - Finding which nucleotide has the highest count from the LacZ gene of E.coli
seq = ("ATGACCATGATTACGGATTCACTGGCCGTCGTTTTACAACGTCGTGACTGGGAAAACCCTGGCGTTACCCAACTTAATCGCCTTGCAGCACATCCCCCTTTCCAGGCCTGCGCTCGCCGCTGCCGGGCGTTTTTTGCGCAGGTGCGGCTACGGCCTGGTGATGCTGCGCCTGCGCCGCGATGTCGCGTTCGGTGCGC")
counts = (seq.count("A"),seq.count("T"),seq.count("G"),seq.count("C"))
bases=["A","T","G","C"]
i=counts.index(max(counts))
print("The nucleotide with the highest count is ", bases[i])
print("The highest count is", counts[i])
```
